### Diagnostic: Anomalous High-|B| Points Inside Holes
Finds |B| spikes above threshold that land inside hole intervals, zooms in on each cluster with +/- 3s context.

```python
# ⭐ Diagnostic: trace anomalous high-|B| points inside holes
# Shows full time series for context, then zoomed clusters for detail
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
import pandas as pd
from plotbot import get_data as plotbot_get_data
from datetime import timezone, timedelta

# Fetch the data for the active time range
plotbot_get_data(trange, mag_rtn_4sa)

# Build a pandas Series from the mag data
bmag_times = mag_rtn_4sa.bmag.datetime_array
bmag_vals = mag_rtn_4sa.bmag.data
bmag_series = pd.Series(bmag_vals, index=pd.DatetimeIndex(bmag_times))
bmag_clipped = bmag_series[trange[0]:trange[1]]

# Parse the marker file to get hole intervals (reuse showda_holes internals)
from plotbot.showda_holes import _parse_marker_file
hole_intervals = _parse_marker_file(mh_marker_file_path, pm)

# Classify each point as inside/outside a hole
is_inside = np.zeros(len(bmag_clipped), dtype=bool)
for i, t in enumerate(bmag_clipped.index):
    t_utc = t.tz_convert('UTC') if t.tz is not None else t.tz_localize('UTC')
    for h_start, h_end in hole_intervals:
        if h_start <= t_utc <= h_end:
            is_inside[i] = True
            break

# Find the anomalous high-bmag points (inside holes AND bmag > threshold)
threshold = 300  # nT -- adjust if needed to catch your 3 points
anomalous_mask = is_inside & (bmag_clipped.values > threshold)
anomalous_times = bmag_clipped.index[anomalous_mask]
anomalous_vals = bmag_clipped.values[anomalous_mask]

print(f"Found {len(anomalous_times)} anomalous points with |B| > {threshold} nT inside holes:")
for t, v in zip(anomalous_times, anomalous_vals):
    print(f"  {t}  |B| = {v:.1f} nT")

# Identify which hole each falls in
for t, v in zip(anomalous_times, anomalous_vals):
    t_utc = t.tz_convert('UTC') if t.tz is not None else t.tz_localize('UTC')
    for hi, (h_start, h_end) in enumerate(hole_intervals):
        if h_start <= t_utc <= h_end:
            duration_ms = (h_end - h_start).total_seconds() * 1000
            print(f"  -> Hole #{hi}: {h_start.strftime('%H:%M:%S.%f')} to {h_end.strftime('%H:%M:%S.%f')} ({duration_ms:.0f} ms)")
            break

# --- PLOT 1: Full time series overview for context ---
fig_overview, ax_ov = plt.subplots(figsize=(16, 5))
ax_ov.plot(bmag_clipped.index, bmag_clipped.values, 'k-', linewidth=0.5, alpha=0.7, label='|B|')

for h_start, h_end in hole_intervals:
    h_start_naive = h_start.replace(tzinfo=None)
    h_end_naive = h_end.replace(tzinfo=None)
    if h_start_naive >= bmag_clipped.index[0] and h_end_naive <= bmag_clipped.index[-1]:
        ax_ov.axvspan(h_start_naive, h_end_naive, alpha=0.15, color='red')

if len(anomalous_times) > 0:
    ax_ov.scatter(anomalous_times, anomalous_vals, c='red', s=100, zorder=5,
                  edgecolors='black', linewidths=1.5,
                  label=f'Anomalous ({len(anomalous_times)} pts, |B| > {threshold} nT)')

ax_ov.set_xlabel('Time (UTC)', fontsize=12)
ax_ov.set_ylabel('|B| (nT)', fontsize=12)
ax_ov.set_title(f'Full Context: |B| Time Series with Holes (red shading) and Anomalous Points\n{trange[0]} to {trange[1]}', fontsize=14)
ax_ov.legend(fontsize=11)
ax_ov.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
ax_ov.grid(True, linestyle=':', alpha=0.5)
fig_overview.tight_layout()
plt.show()

# --- Group anomalous points into clusters (within 5s of each other) ---
clusters = []
if len(anomalous_times) > 0:
    current_cluster = [0]
    for j in range(1, len(anomalous_times)):
        if (anomalous_times[j] - anomalous_times[current_cluster[-1]]).total_seconds() < 5.0:
            current_cluster.append(j)
        else:
            clusters.append(current_cluster)
            current_cluster = [j]
    clusters.append(current_cluster)

print(f"\n{len(clusters)} cluster(s) of anomalous points found.")

# --- PLOT 2: Zoomed subplots per cluster, +/- 3 seconds of context ---
ZOOM_PADDING = timedelta(seconds=3)
n_clusters = len(clusters)
if n_clusters > 0:
    fig, axes = plt.subplots(n_clusters, 1, figsize=(16, 4 * n_clusters), squeeze=False)
    for ci, cluster_indices in enumerate(clusters):
        ax = axes[ci, 0]
        c_times = anomalous_times[cluster_indices]
        c_vals = anomalous_vals[cluster_indices]

        zoom_start = c_times[0] - ZOOM_PADDING
        zoom_end = c_times[-1] + ZOOM_PADDING
        zoom_data = bmag_clipped[zoom_start:zoom_end]

        ax.plot(zoom_data.index, zoom_data.values, 'k-', linewidth=1.0, alpha=0.8)

        for h_start, h_end in hole_intervals:
            h_start_naive = h_start.replace(tzinfo=None)
            h_end_naive = h_end.replace(tzinfo=None)
            if h_end_naive >= zoom_start and h_start_naive <= zoom_end:
                ax.axvspan(max(h_start_naive, zoom_start), min(h_end_naive, zoom_end),
                          alpha=0.2, color='red', label='Hole interval' if ci == 0 else None)

        ax.scatter(c_times, c_vals, c='red', s=150, zorder=5, edgecolors='black',
                  linewidths=2, label=f'{len(c_times)} anomalous pts')

        for t, v in zip(c_times, c_vals):
            ax.annotate(f'{v:.0f} nT', xy=(t, v), xytext=(5, 10),
                       textcoords='offset points', fontsize=10, fontweight='bold', color='red')

        ax.set_ylabel('|B| (nT)', fontsize=12)
        ax.set_title(f'Cluster {ci+1}: {c_times[0].strftime("%H:%M:%S.%f")[:-3]} ({len(c_times)} pts)', fontsize=13)
        ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M:%S'))
        ax.legend(fontsize=10, loc='upper right')
        ax.grid(True, linestyle=':', alpha=0.5)

    axes[-1, 0].set_xlabel('Time (UTC)', fontsize=12)
    fig.suptitle(f'Zoomed View: Anomalous |B| > {threshold} nT Points Inside Holes', fontsize=15, y=1.01)
    fig.tight_layout()
    plt.show()

print(f"\nMedian |B| in range: {np.nanmedian(bmag_clipped.values):.1f} nT")
print(f"Max |B| in range: {np.nanmax(bmag_clipped.values):.1f} nT")
if len(anomalous_vals) > 0:
    print(f"Anomalous points at ~{np.mean(anomalous_vals):.0f} nT = {np.mean(anomalous_vals)/np.nanmedian(bmag_clipped.values):.1f}x the median")
```
